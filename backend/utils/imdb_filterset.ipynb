```python
import pandas as pd
```

```python
df = pd.read_csv("../resources/Imdb Movie Dataset.csv", low_memory=False)
```

```python
print(df.columns)
```

```python
dropped_cols = ['status', 'original_title', 'production_companies', 'release_date', 'tagline', 'budget', 'adult', 'production_countries']
df = df.drop(columns=dropped_cols)

print(df.columns)
```

```python
vote_count_thresh = 5000
vote_avg_thresh = 3.0

filtered_df = df[
    (df['vote_count'] >= vote_count_thresh) &
    (df['vote_average'] >= vote_avg_thresh) 
    #(df['spoken_languages'].str.contains("English", na=False))
    #(df['original_language'].str.contains("en", na=False))
]

filtered_df.head()
```

```python
filtered_df.to_csv('resources/imdb_filtered.csv', index=False)
```
